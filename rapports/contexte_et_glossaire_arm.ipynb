# 1. Contexte et objectif du projet

Le projet **Fighting Maternal Risk** porte sur l’analyse de données relatives à des femmes enceintes et sur l’étude de leur association avec différents niveaux de **risque maternel**.

L’objectif du notebook actuel n’est pas de construire un système médical de diagnostic ou de prédiction individuelle. Le travail consiste principalement à comprendre les données, vérifier leur qualité, nettoyer uniquement les problèmes justifiés, comparer les caractéristiques selon le niveau de risque, visualiser les associations et formuler des conclusions descriptives.

### Problématique

> **Dans quelle mesure les caractéristiques physiologiques et sociodémographiques des femmes enceintes sont-elles associées aux différents niveaux de risque maternel observés dans les données ?**

# 2. Comprendre le domaine

Dans le cadre de ce projet, une observation représente une femme enceinte pour laquelle plusieurs mesures physiologiques et un niveau de risque sont disponibles.

Le jeu de données ne fournit pas une description clinique complète de chaque grossesse. Les variables disponibles sont notamment l’âge, la pression artérielle, la glycémie, la température corporelle, la fréquence cardiaque et le niveau de risque.

Le **risque maternel** désigne ici le niveau associé à l’observation dans le dataset : **Low Risk**, **Mid Risk** ou **High Risk**. Les critères médicaux exacts ayant servi à construire cette classification ne sont pas documentés dans les fichiers étudiés.

La correspondance utilisée dans le notebook est :

| Code | Libellé |
|---:|---|
| 0 | Low Risk |
| 1 | Mid Risk |
| 2 | High Risk |

Cette correspondance est explicitement considérée comme une hypothèse de travail.

# 5. Comprendre les variables médicales

## 5.1 Age

`Age` est l’âge en années. C’est une variable sociodémographique. Le projet observe une plage de **10 à 70 ans**. Ces valeurs atypiques ne sont pas supprimées automatiquement : elles doivent être documentées et analysées. Le notebook prévoit notamment une analyse des 10–14 ans, des 50 ans et plus et des cas de 10, 11 et 70 ans lorsqu’ils existent.

## 5.2 SystolicBP

`SystolicBP` correspond à la pression artérielle systolique, en mmHg. Elle constitue la valeur supérieure de la mesure de pression artérielle. Le projet crée notamment une variable d’analyse selon `SystolicBP >= 140`.

## 5.3 DiastolicBP

`DiastolicBP` correspond à la pression artérielle diastolique, en mmHg. Elle constitue la valeur inférieure de la mesure de pression artérielle.

## 5.4 Blood glucose

La glycémie correspond à la concentration de glucose dans le sang. Elle est exprimée ici en mmol/L. La colonne est renommée `BloodGlucose`. Le projet crée aussi quatre classes : **≤ 7**, **7–9**, **9–12** et **> 12**.

## 5.5 BodyTemp

`BodyTemp` représente la température corporelle brute. Le metadata indique °C, mais les valeurs observées, comprises entre 98 et 103, sont traitées dans le notebook comme des valeurs en Fahrenheit. Une nouvelle variable `BodyTemp_C` est créée avec `(BodyTemp - 32) × 5/9`, arrondie à une décimale.

## 5.6 HeartRate

`HeartRate` représente la fréquence cardiaque, en bpm, c’est-à-dire battements par minute. Deux observations valent **7 bpm** et sont considérées comme des anomalies de saisie dans le cadre du projet. Elles sont retirées avec la règle `HeartRate < 40`.

## 5.7 RiskLevel

`RiskLevel` représente le niveau de risque codé numériquement. Le projet crée `RiskLabel` pour rendre les catégories lisibles.

# 14. Association, causalité et interprétation

Une **association** signifie que deux variables présentent une relation dans les données observées.

La **causalité** signifie qu’une variation d’un phénomène produit un effet sur un autre.

Le projet ne permet pas d’établir la causalité.

Par exemple, si la glycémie présente la corrélation la plus forte avec `RiskLevel`, la formulation correcte est de dire qu’elle présente une association plus forte dans le dataset étudié. Il ne faut pas écrire que la glycémie cause le niveau de risque.

Cette distinction est essentielle pour conserver une interprétation scientifique correcte.

# 15. Limites du projet

Le projet est une analyse descriptive et présente plusieurs limites :

- les données sont observationnelles ;
- les données proviennent de zones rurales du Bangladesh, ce qui limite la généralisation ;
- certaines variables potentiellement importantes sont absentes, notamment parité, IMC et antécédents médicaux ;
- les données sont transversales et ne fournissent pas un suivi longitudinal complet ;
- les critères exacts de classification de `RiskLevel` ne sont pas documentés ;
- la corrélation de Pearson avec une cible ordinale codée 0/1/2 reste une indication descriptive ;
- les âges extrêmes ne sont pas automatiquement déclarés invalides ;
- les seuils utilisés dans les analyses ne doivent pas être présentés comme des recommandations médicales.

# 16. Ce que le projet ne fait pas

Le notebook actuel ne réalise pas :

- de machine learning ;
- de prédiction médicale individuelle ;
- de diagnostic ;
- de conclusion causale ;
- de validation clinique des seuils ;
- de validation externe sur une autre population.

Le machine learning, l’ajout de variables telles que la parité ou l’IMC, le suivi longitudinal et la validation externe sont présentés comme des évolutions futures.

# 19. Comment lire correctement les résultats

Une interprétation rigoureuse suit trois étapes :

### 1. Décrire
Dire ce qui est observé dans les données.

### 2. Interpréter avec prudence
Dire ce que l’observation suggère dans le cadre du dataset.

### 3. Poser la limite
Préciser ce que l’analyse ne permet pas d’affirmer.

Exemple :

> `BloodGlucose` présente une corrélation de 0,56 avec `RiskLevel` dans le dataset étudié. Cela indique une association numérique plus forte que celles observées pour les autres variables analysées. Cette association ne permet pas de conclure que la glycémie cause à elle seule le niveau de risque.

Cette méthode doit être appliquée à toutes les conclusions du projet.
